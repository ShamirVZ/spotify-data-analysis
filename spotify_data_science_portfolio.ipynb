## Distribución Gráfica de Variables Cualitativas

```python
# Seleccionar solo las columnas cualitativas (tipo 'object') que queremos analizar
columnas_cualitativas = ['genero', 'tono', 'modo', 'compas']
# no incluimos "track_name", "track_id", "artist_name" debido a que tienen un umero extramadamente alto de valores unicos, lo
#cual lo hace inviable para mostrarse en gráficos

# Calcular el número de filas y columnas para el subplot
n = len(columnas_cualitativas)
nrows = (n + 1) // 2  # Aproximadamente dos columnas por fila
ncols = 2

# Crear la figura y los subplots
fig, axes = plt.subplots(nrows=nrows, ncols=ncols, figsize=(15, 6 * nrows))
fig.suptitle('Distribución de Variables Cualitativas', fontsize=16)

# Aplanar el array de ejes para facilitar la iteración
axes = np.array(axes).ravel()

# Crear countplots para cada variable cualitativa
for i, col in enumerate(columnas_cualitativas):
    ax = axes[i]
    sns.countplot(data=df, y=col, ax=ax, palette='viridis', order=df[col].value_counts().index)
    ax.set_title(f'Distribución de {col}')
    ax.set_xlabel('Conteo')
    ax.set_ylabel(col)

# Ocultar subplots vacíos si los hay
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Ajustar el layout y el título
plt.show()
```

# Distribución Gráfica de las variables Numéricas

```python
# Seleccionar solo las columnas numéricas
columnas_numericas = df.select_dtypes(include=[np.number]).columns

# Calcular el número de columnas para el subplot
n = len(columnas_numericas)
ncols = 3 # Número deseado de columnas en la cuadrícula
nrows = (n + ncols - 1) // ncols # Calcular dinámicamente el número de filas

# Crear la figura y los subplots
fig, axes = plt.subplots(nrows=nrows, ncols=ncols, figsize=(15, 5*nrows))
fig.suptitle('Distribución de Variables Numéricas', fontsize=16)

# Aplanar el array de ejes para facilitar la iteración
axes = np.array(axes).ravel()

# Crear histogramas para cada variable numérica
for i, col in enumerate(columnas_numericas):
    ax = axes[i]
    sns.histplot(df[col], ax=ax, bins=50, kde=True)
    ax.set_title(f'Distribución de {col}')
    ax.set_xlabel(col)
    ax.set_ylabel('Frecuencia')
    ax.grid(True, linestyle='--', alpha=0.7) # Añadir rejilla

# Ocultar subplots vacíos si los hay
for j in range(i+1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # Ajustar el layout para el suptitle
plt.show()
```

## Deteción de outliers con la variable "Popularidad"

```python
import matplotlib.pyplot as plt
plt.boxplot(df["popularidad"], patch_artist=True, boxprops=dict(facecolor='skyblue'), medianprops=dict(color='red'))
plt.ylabel("Nivel de Popularidad")
plt.title("Diagrama de Boxplot de Popularidad (Identificación de Outliers)")
plt.show()
```

```python
import seaborn as sns
# Estadísticas necesarias para reproducir el gráfico de cajas y bigotes
Q1 = df['popularidad'].quantile(0.25)
Q3 = df['popularidad'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Estadísticas para popularidad:")
print(f"Q1 - 1.5IQR = {lower_bound:.2f}")
print(f"Q1 = {Q1:.2f}")
print(f"Mediana = {df['popularidad'].median():.2f}")
print(f"Q3 = {Q3:.2f}")
print(f"Q3 + 1.5IQR = {upper_bound:.2f}")
print(f"Número de observaciones: {len(df['popularidad'])}")
print(f"Número de outliers: {sum((df['popularidad'] < lower_bound) | (df['popularidad'] > upper_bound))}")
```

```python
mediana_popularidad = df['popularidad'].median()

print(f"Mediana de popularidad: {mediana_popularidad:.2f}")
#La linea roja de la caja del gráfico de boxplot representa la mediana de mis datos, se observa que la variable "popularidad" presenta outliers
```

# Fundamentación del Tratamiento de Outliers
- La variable "instrumental" presenta outliers (22.15%), pero no deben tratarse ya que representan canciones puramente instrumentales o géneros como la música clásica y el jazz. Eliminarlos causaría un sesgo donde el dataset solo reconocería canciones con contenido vocal, perdiendo la diversidad de géneros no comerciales.

- La variable "hablado" presenta outliers (14.17%), pero deben mantenerse porque identifican pistas con alto contenido de palabra hablada, como podcasts, poemas o sketches de comedia. Su presencia es vital para que el modelo pueda diferenciar entre música y contenido informativo o hablado.

- La variable "en_vivo" presenta outliers (9.13%), pero no se consideran errores, sino que indican grabaciones realizadas en vivo frente a una audiencia. Estos valores son registros reales de la naturaleza de la producción del audio y eliminarlos impediría identificar versiones "Live" de las canciones.

- La variable "popularidad" presenta outliers mínimos (0.02%), pero son precisamente los "Mega-Hits" mundiales. Dado que en la industria musical el éxito es exponencial, estos valores extremos son los datos más valiosos para entender las tendencias de consumo masivo y no deben ser recortados.

- Las variables "bailabilidad", "energia", "acustica" y "valencia" presentan una distribución acotada entre 0 y 1. Aunque estadísticamente algunos valores se alejen de la media, al estar limitados físicamente por un rango estricto, no tienen la capacidad de distorsionar los cálculos matemáticos como lo haría una variable sin límites (como el precio o la edad).

- La variable tempo presenta muy pocos outliers (0.07%), pero estos
corresponden a géneros con ritmos extremadamente rápidos (como el Metal o Drum and Bass) o muy lentos (Ambient). Mantenerlos permite que el análisis cubra todo el espectro de velocidades rítmicas existentes en la música.
